# Boolean Masking

```python
'''
Boolean Indexing--> It is a phenomenom of extracting the values from the
            array based on boolean condition.

Syntax--> arr[condition]

-> Incase of 2D boolean indexing or 3D boolean indexing result willbe in
  1D.


'''
```

```python
sales=np.array([12000,34000,90000,2000,6000,11000,160000])
sales
#Sales above 10000
sales[sales>10000]


#sales greater than 2000 and less than 15000
sales[(sales>2000) & (sales<15000)]

#sales greater than 7000 and less than 20000
sales[(sales>7000)&(sales<20000)]

#sales greater than 10k or less than 90000
sales[(sales>10000)|(sales<90000)]

num=np.array([34,23,11,76,93,57,73,87,50,92])
#values greater than 50
num[num>50]

#values greater than 2 or less than 100
num[(num>2) | (num<100)]
#even numbers
num[num%2==0]

#odd numbers and multiple of 3
num[(num%2!=0) & (num%3==0)]

#even numbers 10-80
num[(num%2==0) & ((num>10) &(num<=80))]

#numbers between 30-80
num[(num>30) & (num<=80)]

```

```python
'''
Employee Salary Analysis


salary = np.array([
25000,
32000,
28000,
45000,
52000,
38000,
60000,
42000,
30000,
55000
])


Find the number of employees.


Find total salary.


Find average salary.


Find minimum salary.


Find maximum salary.

Find employees earning above ₹40,000.


Find employees earning below ₹30,000.


Find salaries between ₹30,000 and ₹50,000.


Count employees earning above ₹40,000.


Find salaries greater than the average.

Calculate the percentage of employees earning above ₹40,000.


Give every employee a 10% increment.

Identify salaries significantly above the average.
'''
salary = np.array([
25000,
32000,
28000,
45000,
52000,
38000,
60000,
42000,
30000,
55000
])

# Find the number of employees.
salary.size
# Find total salary.
salary.sum()

# Find average salary.
salary.mean()

# Find minimum salary.
salary.min()

# Find maximum salary.
salary.max()

# Find employees earning above ₹40,000.

salary[salary>40000]

# Find employees earning below ₹30,000.
salary[salary<30000]

# Find salaries between ₹30,000 and ₹50,000.
salary[(salary>30000)&(salary<50000)]
# Count employees earning above ₹40,000.
salary[salary>40000].size

# Find salaries greater than the average.
salary[salary>salary.mean()]
# Calculate the percentage of employees earning above ₹40,000.
(salary[salary>40000].size/salary.size)*100
# Give every employee a 10% increment.
salary+salary*0.10
# Identify salaries significantly above the average.
salary[salary>salary.mean()]
```
