# series
* it is a 1-Dimensional label data structure.
* it has a single column of data.
* it is like a 1D array.


```python
#creating a series
#syntax--> pd.series(1D array coll,name='col name')

snacks=pd.Series(['sandwich','momos','shawarma','samosa','bhelpuri','kerala chips','chowmein','mancurian'],name="snacks",index=[1,2,3,4,5,6,7,8])
snacks
```

```python
#Data cleaning steps
'''
1. Rename the columns
Syntax-->
    data={old:new}
    df.rename(columns=data)

2. dropping duplicates
syntax--> df.drop_duplicates()--> removes the duplicates values presents in df
--> if we want to remove duplicates based on any specific column
syntax--> df.drop_duplicates(subset=col)

3. changing values present in a column
syntax-->
          data={old:new}
          df[col]=df[col].map(data)
  --> changes the respective values with the new value in a column
4. applying inbuilt functions on a column
syntax-->
          df[col].str.func()

5. handling missing values

    a. removing null values
    syntax--> df.dropna() #removes all the values records from the dataset
    if we want to remove null records based on a specific col
    syntax--> df.dropna(subset='col')

NOTE==> we should remove null values only when it is less than 2.5% of the dataframe

    b. filling null values with any specifc value
    syntax--> df.fillna(mean/mode/median/specific value)

    c. filling null values by backward or forward fill
    syntax--> df[col].ffill()  # add previous data above null value
              df[col].bfill()  # add next data below the null value

6. rearranging the cols
syntax--> df=df[new col order]

7. resetting the index of dataframe
  syntax--> df.reset_index()

8. Dropping unwanted column
  syntax--> df.drop(col,axis=1)

9.dropping unwanted rows
  syntax--> df.drop([1,2,3,4])

10. setting a column as index
  syntax--> df.set_index(col)
'''
```

```python
'''
JOINS
-> It is used to combine two or more dataframes based on common column or index.

-> In real time datas are stored in different tables so joins plays an important role in combining multiple tables.

TYPES OF JOINS
INNER JOIN
OUTER JOIN
LEFT JOIN
RIGHT JOIN
CROSS JOIN

To perform joins we use some methods:-
    concat()--> used to concat two dataframes
    merge()--> used to merge two dataframes on common column

    '''

#concat()
#syntax--> pd.concat(df1,df2,axis=)
```

```python
employees = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry"],
    "Dept_ID": [1, 2, 1, 3, 2, 4, 5, 6],
    "Salary": [60000, 55000, 72000, 65000, 58000, 70000, 52000, 68000]
})
```

```python
departments = pd.DataFrame({
    "Dept_ID": [1, 2, 3, 4, 7],
    "Department": [
        "HR",
        "Finance",
        "IT",
        "Marketing",
        "Legal"
    ],
    "Location": [
        "Hyderabad",
        "Bangalore",
        "Chennai",
        "Mumbai",
        "Delhi"
    ]
})
```

```python
pd.concat((employees,departments),axis=1)
```

```python
pd.concat((employees,departments),axis=0)
```

```python
#merge()
#syntax--> pd.merge(left,right,on='common col',how='join)
pd.merge(employees,departments,on='Dept_ID',how='inner')
```

```python
#inner join
#-> it joins the common records from both left and right table
#syntax--> pd.merge(left,right,on='common col',how='inner')

pd.merge(employees,departments,on='Dept_ID',how='inner')
```

```python
#outer join
#--> it joins both dataframe based on matched as well as unmatched records
#syntax--> pd.merge(left,right,on='common col',how='outer')
```

```python
pd.merge(employees,departments,on='Dept_ID',how='outer')
```

```python
# left join
#--> it will join all the records of left dataframe and join all
# the matched records for the right dataframe

#syntax--> pd.merge(left,right,on='common col',how='left')
pd.merge(employees,departments,on='Dept_ID',how='left')
```

```python
#right join
#--> it will join all the records of right dataframe and join all
# the matched records for the left dataframe

#syntax--> pd.merge(left,right,on='common col',how='left')
pd.merge(employees,departments,on='Dept_ID',how='right')
```

```python
#cross join
#-> it will join both dataframe by every possible combination
#syntax--> pd.merge(left,right,how='cross')

pd.merge(employees,departments,how='cross')
```
