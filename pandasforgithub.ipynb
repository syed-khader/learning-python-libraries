# series
* it is a 1-Dimensional label data structure.
* it has a single column of data.
* it is like a 1D array.


```python
#creating a series
#syntax--> pd.series(1D array coll,name='col name')

snacks=pd.Series(['sandwich','momos','shawarma','samosa','bhelpuri','kerala chips','chowmein','mancurian'],name="snacks",index=[1,2,3,4,5,6,7,8])
snacks
```

```python
# create two series
# one list of your friends
# list of destinations you want to visit before you die

friends=pd.Series(['asim','atif','amir','harsh','shahank','jhon'],name="friends",index=[1,2,3,4,5,6])
friends

list_of_places=pd.Series(['kashmir','goa','sweden','amarica','japan',45],name="places",index=[1,2,3,4,5,6])
list_of_places
```

```python
# Accessing values from series
#syntax--> var[index]
list_of_places[2]

#accessing multiple index position
list_of_places[[2,3,1]]

#access multiple values in a seq manner--> slicing
#syntax--> var[si:ei+-1:step]
list_of_places[2:4]
```

```python
snacks.ndim
snacks.size
snacks.shape
snacks.dtype

# we can perform all the attributes if array on series
```

```python
# difference between series and array
'''
        array                                     series
1. it belongs to numpy                  1. it belongs to pandas
2. string is treated as string          2. string is treated as object
3. np.array(coll)                       3. pd.Series(coll)
4. we can create N dimension arrays     4. it can be only 1D
5. it can only store homogenous         5. can store heterogenous data
'''
```

## DataFrame

-> It is a 2-Dimensional datastructure of pandas.

-> It stores the data in a tabular manner having rows and columns.

-> It is a collection series


```python
#Creating a dataframe
#Syntax-->pd.DataFrame(Dictionary)

df=pd.DataFrame({'Name':['Rahul','Raj','Harish','Sunil','Ganesh'],
                 'City':['KOL','MP','DEL','BBSR','MUM'],
                 'Age':[23,45,24,21,67]})
df

#Creating dataframe using nested list
#Syntax-->pd.DataFrame(data,columns=[col1,col2,...coln])

data=[['Manchurian','full',120],
      ['Chowmein','half',90],
      ['Samosa','half',20],
      ['Egg puff','full',40],
       ['Bonda',None,45]]

df1=pd.DataFrame(data,columns=['Item','Quantity','Price'])
df1
```

```python
prod=pd.DataFrame({'product': ['mobile phone','laptop','AC','washing machine','TV','air fryer','kettle'],'price':[35000,60000,50000,10000,60000,4000,1000]})
prod

# adding a new column
#syntax--> df[new_col]=value

prod['category']='electronics'
prod
prod['category']=['phones','electronics','appliance','appliance','appliance','appliance','appliance']
prod


#dropping a column
#syntax--> df.drop(col,axis=1,inplace=True)
prod.drop('category',axis=1,inplace=True)
prod

#inplace =True will modify the orignal dataframe/data source

#dropping a particular row using index
#syntax-> df.drop(index)
elec=prod.drop(3)
elec

#rename a column
# var={'old col':'new col'}
#df.rename(columns=var)

data={'price':'MRP'}
prod.rename(columns=data,inplace=True)
prod
```
