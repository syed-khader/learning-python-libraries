```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
from matplotlib.lines import lineMarkers
arr1=np.array([1,2,3,4])
arr2=arr1**2

plt.plot(arr1,arr2,marker='o',markerfacecolor='red',markeredgecolor='red',linewidth=3,color='red',linestyle='dotted',label='X,Y') # s= square , d=dimond , +,*,.,o,p=pentagon,>,<
plt.legend()
plt.grid()
plt.title('squre graph')
plt.xlabel('values of x',fontsize=14)
plt.ylabel('squares of y',fontsize=14,color='blue')

# marker --> (used to select coordinate symbols)# s= square , d=dimond , +,*,.,o,p=pentagon,>,<
#markerfacecolor ( used to color the marker)
#markeredgecolor ( adjust the line width in plot)
#linewidth ( adjust the line width in plot)
# linestyle (adjust the style of line)--> -,--,-.,., dotted,solid,dashed
#plt.title()--> used to give a title to the plot
# syntax--> plt.title('title')
#plt.xlabel()--> used to give label to x axis
# plt.xlabel('values of x',fontsize=14)

```

# line plot
1. it is used to connect the data
point in a line
2. it is usually used to show trends and changes across the sequences.
3. monthly sales , temp , variation , stock prices, etc

4. syntax--> plt.plot(x,y)


```python

arr3=np.array([1,2,3,4,5,6])
arr4=arr3**3

plt.plot(arr3,arr4,marker='o',markerfacecolor='red',markersize=3,markeredgecolor='red',linewidth=1,label='x cube y',linestyle='dashed')
plt.legend()
plt.grid()
plt.title('cube graph')
plt.xlabel('values of x',fontsize=14)
plt.ylabel('cube of y',fontsize=14,color='blue')
plt.show()
```

```python
df1=pd.DataFrame({'Product':['Mobile','TV','AC','Laptop','Washing machine','Oven'],
                  'Sales':[20000,40000,50000,65000,20000,10000]})
plt.plot(df1['Product'],df1['Sales'],markerfacecolor='red',marker='.',color='black',markersize=8,markeredgecolor='red')
plt.xticks(rotation=45)
```

#pie chart

1. it is used to show the percentage values distributed of a numerical column

syntax--> plt.pie(data)

```python
marks=[90,34,67,56,23]
sub=['physics','chemistry','maths','biology','english']
plt.pie(marks,labels=sub,autopct='%0.2f%%',colors=['red','orange','blue','green','purple'],explode=[0.1,0,0,0,0],shadow=True)
plt.title('marks distribution over subject')
plt.show()

```

```python
# plot percentage of passenger for male and female
dft.groupby('sex').count()
gender=['Female','Male']
plt.pie(dft.groupby('sex')['name'].count(),labels=gender,colors=['pink','blue'],autopct='%0.2f%%')
plt.title('percentage of passenger for male and female')
plt.show()

```

```python
#plot percentage of logins in each department
# df2.groupby()
#plot percentage of people who died and survived in titanic
survived=dft['survived'].value_counts()
l=['died','survived']
plt.pie(survived,labels=survived.index,autopct='%0.2f%%',colors=['gray','yellow'],startangle=90)
plt.title('percentage of people who died and survived in titanic ')
plt.show()
```

```python
#plot the percentage of item type in tier 1 city
df2
tire_1=df2[df2['Outlet Location Type']=='Tier 1']
types=tire_1['Item Type'].value_counts()
plt.pie(types,labels=types.index,autopct='%0.2f%%')
plt.title('percentage of item type in tier 1 city')
plt.show()
```

# seaborn
1. it is a open source library used for advanced visualization
2. seaborn is based on matplotlib
3. to use seaborn styling we need to install library
# syntax--> pip install seaborn as sns


```python
'''
         matplotlib                                                    seaborn
      -------------------------------------------------------------------------------------------
      1. has limited plots                               1. more plots
      2. manual styling                                  2. automatic styling
      3. it is not based on any visualization            3. it is based on matplotlib

    -> there are types of plots in seaborn
        1. distribution plots
        2. categorial plots
        3. matrix plot
        4. regression plot

    -> based on the columns plots are divided into
        1. univariate --> plotting based on a single column
        2. bivariate --> plotting based on 2 columns
        3. multivariate --> plotting based on more then 2 columns
        '''
```

```python
'''
DISTRIBUTION PLOTS
-------------------
-> it shows the distribution of the data
-> it works on numerical columns
-> it helps to understand statistical information of the data
'''
```
