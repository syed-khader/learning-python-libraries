# matplotlib
1. it is a standard visualization library of python.
2. to use the plots of matplotlib we need to install it.
3. syntax--> pip install matplotlib
4. after installing we need to import it
5.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
from matplotlib.lines import lineMarkers
arr1=np.array([1,2,3,4])
arr2=arr1**2

plt.plot(arr1,arr2,marker='o',markerfacecolor='red',markeredgecolor='red',linewidth=3,color='red',linestyle='dotted',label='X,Y') # s= square , d=dimond , +,*,.,o,p=pentagon,>,<
plt.legend()
plt.grid()
plt.title('squre graph')
plt.xlabel('values of x',fontsize=14)
plt.ylabel('squares of y',fontsize=14,color='blue')

# marker --> (used to select coordinate symbols)# s= square , d=dimond , +,*,.,o,p=pentagon,>,<
#markerfacecolor ( used to color the marker)
#markeredgecolor ( adjust the line width in plot)
#linewidth ( adjust the line width in plot)
# linestyle (adjust the style of line)--> -,--,-.,., dotted,solid,dashed
#plt.title()--> used to give a title to the plot
# syntax--> plt.title('title')
#plt.xlabel()--> used to give label to x axis
# plt.xlabel('values of x',fontsize=14)

```

# line plot
1. it is used to connect the data
point in a line
2. it is usually used to show trends and changes across the sequences.
3. monthly sales , temp , variation , stock prices, etc

4. syntax--> plt.plot(x,y)


```python

arr3=np.array([1,2,3,4,5,6])
arr4=arr3**3

plt.plot(arr3,arr4,marker='o',markerfacecolor='red',markersize=3,markeredgecolor='red',linewidth=1,label='x cube y',linestyle='dashed')
plt.legend()
plt.grid()
plt.title('cube graph')
plt.xlabel('values of x',fontsize=14)
plt.ylabel('cube of y',fontsize=14,color='blue')
plt.show()
```

```python
df1=pd.DataFrame({'Product':['Mobile','TV','AC','Laptop','Washing machine','Oven'],
                  'Sales':[20000,40000,50000,65000,20000,10000]})
plt.plot(df1['Product'],df1['Sales'],markerfacecolor='red',marker='.',color='black',markersize=8,markeredgecolor='red')
plt.xticks(rotation=45)
```
